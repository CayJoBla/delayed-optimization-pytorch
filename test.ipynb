```python
from torch import nn
from torch.optim import Adam
from delay_optimizer import DelayedOptimizer
from delay_optimizer.delays.distributions import Stochastic

model = nn.Sequential(
    nn.Linear(1000,1000),
    nn.ReLU(),
    nn.Linear(1000,1000),
    nn.ReLU(),
    nn.Linear(1000,1000),
    nn.ReLU(),
    nn.Linear(1000,1000),
    nn.ReLU(),
    nn.Linear(1000,1000),
    nn.ReLU(),
    nn.Linear(1000,1000),
    nn.ReLU(),
    nn.Linear(1000,1000),
    nn.ReLU(),
    nn.Linear(1000,1000),
    nn.ReLU(),
    nn.Linear(1000,1000),
    nn.ReLU(),
    nn.Linear(1000,1000),
    nn.ReLU(),
    nn.Linear(1000,1000),
    nn.ReLU(),
    nn.Linear(1000,1000),
    nn.ReLU(),
    nn.Linear(1000,1000),
    nn.ReLU(),
    nn.Linear(1000,1000),
)
model.to("cuda")
delay = Stochastic(max_L=1)
optimizer = DelayedOptimizer(Adam)(model.parameters(), lr=0.001, delay=delay, history_device="cpu")
```

```python
import torch

delay_times = []
for i in range(1000):
    x = torch.randn(10,1000).to("cuda")
    y = torch.randint(0,1000,(10,)).to("cuda")

    delay_times.append(optimizer.zero_grad())
    loss = nn.CrossEntropyLoss()(model(x),y)
    loss.backward()
    optimizer.step()
```
